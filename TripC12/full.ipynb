```python
data1 = np.loadtxt("./Data-20262506/01.dat")
data2 = np.loadtxt("./Data-20262506/02.dat")
data3 = np.loadtxt("./Data-20262506/03.dat")
data4 = np.loadtxt("./Data-20262506/04.dat")
data5 = np.loadtxt("./Data-20262506/05.dat")
data6 = np.loadtxt("./Data-20262506/06.dat")
data7 = np.loadtxt("./Data-20262506/07.dat")
data8 = np.loadtxt("./Data-20262506/08.dat")
data9 = np.loadtxt("./Data-20262506/09.dat")
data10 = np.loadtxt("./Data-20262506/10.dat")
data11 = np.loadtxt("./Data-20262506/11.dat")



dataset = [
    data1, 'Data 1', 'red',
    data2, 'Data 2', 'blue',
    data3, 'Data 3', 'green',
    data4, 'Data 4', 'orange',
    data5, 'Data 5', 'purple',
    data6, 'Data 6', 'brown',
    data7, 'Data 7', 'pink',
    data8, 'Data 8', 'cyan',
    data9, 'Data 9', 'gray',
    data10, 'Data 10', 'black',
    data11, 'Data 11', 'magenta'
]

for data, label, color in zip(dataset[::3], dataset[1::3], dataset[2::3]):
    V_GS = data[:, 0]
    I_DS = data[:, 2]
    V_DS = data[:, 3]
    # V_C = 0.5 * (data[:, 4] + data[:, 5])
    V_12 = data[:, 6]
    sigma = -data[:, 7]

    plt.plot(
        -V_GS,
        sigma * 1e6,
        color=color,
        label=label
    )



plt.xlabel(r"Gate Voltage -$V_\mathrm{G4}$ (V)")
plt.ylabel(r"Conductance $\sigma$ ($\mu$S/sq)")
plt.xlim(30, 50)
plt.ylim(0, 1)
plt.title("Conductance vs. Gate Voltage")
plt.show()

C = 1.8e-9
A = 32e-6
Ci = C / A
```

```python

for data, label, color in zip(dataset[::3], dataset[1::3], dataset[2::3]):
    V_GS = data[:, 0]
    I_DS = data[:, 2]
    V_DS = data[:, 3]
    # V_C = 0.5 * (data[:, 4] + data[:, 5])
    V_12 = data[:, 6]
    sigma = -data[:, 7]
    mu_var = -sigma/Ci/V_GS * 1e4

    plt.plot(
        -V_GS,
        mu_var,
        color=color,
        label=label
    )

plt.xlabel(r"Gate Voltage -$V_\mathrm{G4}$ (V)")
plt.ylabel(r"Mobility ($\mu$ (cm$^2$/Vs))")
plt.title("Mobility vs. Gate Voltage")
plt.xlim(30, 50)
plt.ylim(0, 5)
# plt.legend()
# plt.savefig("mu_cold.eps", format = 'eps')
plt.show()

```
