```python
data = np.loadtxt('Data-20260705/20260507-0001.csv', delimiter=',', skiprows=1)
time = data[:, 0]
V_in = data[:, 1]
V_out = data[:, 2]

window_size = 5
window = np.ones(window_size) / window_size

V_in_smooth = np.convolve(V_in, window, mode='same')
V_out_smooth = np.convolve(V_out, window, mode='same')

plt.plot(time, -V_in_smooth, label='Input Voltage', color='blue', lw=1.5)
plt.plot(time, 0.1+V_out_smooth/1000*41, label='Output Voltage', color='red', lw=1.5)
plt.xlabel('Time (ms)', fontsize=18)
plt.ylabel('Voltage (V)', fontsize=18)
plt.xlim(-10, 10)
plt.ylim(-2, 2)
plt.title('PS9 Amplifier Response', fontsize=20)
plt.legend()
plt.show()
```

```python
data = np.loadtxt('Data-20260805/20260508-0002.csv', delimiter=',', skiprows=1)
time = data[:, 0]
V_in = data[:, 1]
V_out = data[:, 2]

window_size = 1
window = np.ones(window_size) / window_size

V_in_smooth = np.convolve(V_in, window, mode='same')
V_out_smooth = np.convolve(V_out, window, mode='same')

dt = np.mean(np.diff(time))
freq = np.fft.rfftfreq(len(time), d=dt)

window = np.hanning(len(time))
window_scale = 2.0 / np.sum(window)

V_in_fft = np.fft.rfft(V_in_smooth * window)
V_out_fft = np.fft.rfft((0.3 + 2 * V_out_smooth) * window)


V_in_mag = window_scale * np.abs(V_in_fft)
V_out_mag = window_scale * np.abs(V_out_fft)

gain_mask = (freq > 0) & (V_in_mag > V_in_mag.max() * 1e-3)
gain_db = 20 * np.log10(np.maximum(V_out_mag[gain_mask] / V_in_mag[gain_mask], 1e-12))

fig, axes = plt.subplots(2, 1, figsize=(9, 7), sharex=True)

axes[0].semilogx(freq, np.maximum(V_in_mag, 1e-15), label='Input spectrum', color='blue', lw=1.5)
axes[0].semilogx(freq, np.maximum(V_out_mag, 1e-15), label='Output spectrum', color='red', lw=1.5)
axes[0].set_ylabel('Magnitude (V)')
axes[0].set_title('PS9 Amplifier Fourier Analysis')
axes[0].legend()

axes[1].semilogx(freq[gain_mask], gain_db, marker='^', color='black', lw=1.5, markersize=1)
axes[1].set_xlabel('Frequency (MHz)')
axes[1].set_ylabel('Gain (dB)')
axes[1].set_xlim(1e-2, 1e0)
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```
